```python
import seaborn as sns
from matplotlib import pyplot as plt
df=sns.load_dataset('tips')
sns.histplot(df['total_bill'],kde=True)  # Density Curve ko kde
plt.show()
```

```python
#Outlier Detection
sns.boxplot(x='day',y='total_bill' ,data=df)
plt.show()
```

```python
#Advance Distribution
sns.violinplot(x="sex",y="total_bill",data=df)
plt.grid()
plt.show()
```

```python
#Stastical Aggrigation ke liyea  Bar plot ka use hota hai
sns.barplot(x='day',y='total_bill',data=df)
 
```

```python
#Correlation Matrix
corr=df.corr(numeric_only =True)
sns.heatmap(corr,annot=True,cmap='summer')
plt.show()
```

```python
sns.pairplot(df,hue='sex') 
plt.show()
```

```python
sns.regplot(x='total_bill', y='tip',data=df)
plt.show()
```

```python
sns.countplot(x='day',data=df)
plt.show()
```

```python
sns.set_style('darkgrid')
sns.set_palette('Set2')
sns.boxplot(x='day',y='total_bill',data=df)
plt.show()
```

```python
#Main Teachin is starting from here Computer Vision pe cv means
import cv2

# Image read karo
img = cv2.imread('anime.png')
cv2.line(img,(0,0),(200,0),(255,0,0),8)
cv2.rectangle(img,(50,50),(200,200),(0,0,255),-1)
cv2.circle(img,(100,150),50,(0,255,0),-1)

cv2.imshow("Anime Image",img)  # here we can change variable lik gray or Image 
cv2.waitKey(0)
cv2.destroyAllWindows()   #it will clean all memory 
```
