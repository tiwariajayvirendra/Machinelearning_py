```python
from transformers import pipeline
```

```python
generator=pipeline("text-generation",model="openai-community/gpt2")
result=generator("Artificial Intelligence is",max_length=50)
print(result[0]["generated_text"])
```

```python
summarizer=pipeline('summarization')
text="""Artificial Intelligence is a field dominated by the research 
community, and it's important to understand that it is not the future 
of artificial intelligence. There is a certain level of sophistication 
and sophistication built into artificial intelligence technology, and 
that sophistication is often associated with the ability to quickly
identify, diagnose, and respond to problems from within the environment.
For instance, in an effort to detect drug resistance, you need to 
know where drugs are and where they are not, what the risk is and
how to stop them."""
summary=summarizer(text)
print(summary[0]["summary_text"])
```

```python
from transformers import pipeline

summarizer = pipeline("summarization", model="facebook/bart-large-cnn")
text = "Artificial Intelligence is a field dominated by the research community..."
summary = summarizer(text, max_length=100, min_length=30, do_sample=False)
print(summary[0]["summary_text"])
```
