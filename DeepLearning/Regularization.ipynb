```python
from tensorflow.keras import layers,regularizers
from tensorflow.keras.layers import Dense, Dropout
from tensorflow.keras.models import Sequential
from tensorflow.keras.datasets import mnist
```

```python
(X_train,y_train),(X_test,y_test)=mnist.load_data()
```

```python
model =Sequential()
```

```python
#L1. Regularization Loss

model.add(Dense(64,activation='relu',kernel_regularizer=regularizers.l1(0.001)))
```

```python
#L2 Regularization 
model.add(Dense(64,activation='relu',kernel_regularizer=regularizers.l2(0.001)))
```

Regularization

Regularization is a set of technique used to reduce overfitting and improve 
a models ability to genralize unseen data.

What is Overfitting?

Overfitting occurs when a model learns the training data too well, including its noise 
and outliers. if models learns overfitted then 

1. Training accuracy is High
2. Testing accuracy is Low

Regularization help prevent this by making the model less complex or by  
discouraging it from relying too heavy on specific feature.

Types of Regularization

1.)L1 Regularization(Lasso)
  L1 regularization adds the absolute values of the weight to the loss function.
  Formula
  =======
  Loss=Original Loss+ sigma weight

2.)L2 Regualrizationn (Ridge)
L2 Reularization adds the squared vales of the weith to the loss function 

Loss = Original Loss + Sigma  Weith^2



3).Dropout
Dropout randomly Deactivates (drops) a percentage of neurons during trainging
Example(100 Neurons
Dropouts = .5
About 50 Neurons are Randomly Ignored in each trainging step.
benifits of dropout:= Prevent neurons from becoming too dependent on each other.
improves generatlization


4)Early stopping 
Ans. Training stops autometically when the validation  loss stop imporving


5) Batch Normalization
   Batch Noralization normalizes the output of a layers during training
   .
   Benifits:
       1.Faster Convergence
       2.More stable Tringing
       3.Can Provide a regularization effect.
Purpose of Regularization

L1.features Selection

```python
model.add(Dropout(.5))
```

```python
from tensorflow.keras.callbacks import EarlyStopping
```

```python
early_stop=EarlyStopping(monitor='val_loss',patience=5,restore_best_weights=True)
```

```python
model.fit(X_train,y_train,epochs=100,validation_data=(X_test,y_test),callbacks=[early_stop])
```
