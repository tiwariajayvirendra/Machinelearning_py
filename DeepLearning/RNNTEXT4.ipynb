```python
# Text Based RNN
import numpy as np
import tensorflow as tf
```

```python
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import SimpleRNN,Dense,Embedding
```

```python
text=["i love machine learning","i love deep learing","python code","deep learing is powerful","ai is like human"]
```

```python
tokenizer=Tokenizer()
```

```python
tokenizer.fit_on_texts(text)
```

```python
tokenizer
```

```python
total_words=len(tokenizer.word_index)+1
```

```python
total_words
```

```python
# input_sequence=[]
# for line in text:
#     token_list=tokenizer.texts_to_sequences([line])[0]
#     for i in range(1,len(token_list)):
#         n_gram=token_list[:i+1]
# input_sequence.append(n_gram)
# input_sequence


input_sequence = []

for line in text:
    token_list = tokenizer.texts_to_sequences([line])[0]
    for i in range(1, len(token_list)):
        n_gram = token_list[:i+1]
        input_sequence.append(n_gram)   # 👈 यह line अंदर होनी चाहिए
```

```python
#Pad sequence

max_len=max(len(seq) for seq in input_sequence)
input_sequence=pad_sequences(input_sequence,maxlen=max_len,padding='pre')
```

```python
#Split Inot X and Y
X=input_sequence[:, :-1]
y=input_sequence[:, -1]
y
```

```python
model=Sequential([Embedding(total_words,10,input_length=max_len-1),SimpleRNN(32),Dense(total_words,activation='softmax')])
```

```python
model.compile(loss='sparse_categorical_crossentropy',optimizer='adam',metrics=['accuracy'])
```

```python
#Train
model.fit(X,y,epochs=300,verbose=1)
```

```python
# Prediction Next 
seed_text="python"
sequence=tokenizer.texts_to_sequences([seed_text])[0]
sequence=pad_sequences([sequence],maxlen=max_len-1,padding='pre')
prediction=model.predict(sequence,verbose=1)
predicted_word=""
for word,index in tokenizer.word_index.items():
    if np.argmax(prediction)==index:
        predicted_word=word
        break
print("Input",seed_text)
print("Next Word",predicted_word)
```
