```python
import cv2

# Haarcascade model load karo
face = cv2.CascadeClassifier(r"C:\Users\ajayt\OneDrive\Desktop\TrainedModel\haarcascade_frontalface_default.xml")

# Camera open karo
cam = cv2.VideoCapture(0)

while True:
    ind, frame = cam.read()
    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

    # Detect faces
    cascade = face.detectMultiScale(gray, 1.3, 5)

    # Draw rectangle around faces
    for (x, y, w, h) in cascade:
        cv2.rectangle(frame, (x, y), (x + w, y + h), (20, 34, 200), 2)

    # Show video frame
    cv2.imshow("Video", frame)

    # ESC key (27) press karne par break
    if cv2.waitKey(1) == 27:
        break

cam.release()
cv2.destroyAllWindows()
```

```python
import cv2

# Haarcascade model load karo
face = cv2.CascadeClassifier(r"C:\Users\ajayt\OneDrive\Desktop\TrainedModel\haarcascade_frontalface_default.xml")
eyes=cv2.CascadeClassifier(r"C:\Users\ajayt\OneDrive\Desktop\TrainedModel\haarcascade_eye.xml")
# Camera open karo
cam = cv2.VideoCapture(0)

while True:
    ind, frame = cam.read()
    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

    # Detect faces
    cascade = face.detectMultiScale(gray, 1.3, 5)

    # Draw rectangle around faces
    for (x, y, w, h) in cascade:
        cv2.rectangle(frame, (x, y), (x + w, y + h), (20, 34, 200), 2)
        roi_gray=gray[y:y+h,x:x+w]
        roi_color=frame[y:y+h,x:x+w]
        eyecascade=eyes.detectMultiScale(roi_gray)
        for (ex,ey,ew,eh) in eyecascade:
            cv2.rectangle(roi_color,(ex,ey),(ex+ew,ey+eh),(0,255,0),4)
        # Show video frame 
        cv2.imshow("Video", frame)

    # ESC key (27) press karne par break
    if cv2.waitKey(1) == 27:
        break

cam.release()
cv2.destroyAllWindows()
```

```python
import cv2
img=cv2.imread(r"C:\Users\ajayt\Downloads\paint.jpg")
gray=cv2.cvtColor(img,cv2.COLOR_BGR2GRAY)
_,thresh=cv2.threshold(gray,127,255,cv2.THRESH_BINARY)
countours,_=cv2.findContours(thresh,cv2.RETR_TREE,cv2.CHAIN_APPROX_SIMPLE)
for cnt in countours:
    x,y,w,h=cv2.boundingRect(cnt)
    cv2.rectangle(img,(x,y),(x+w,y+h),(0,255,0),2)

cv2.imshow("Object",img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
