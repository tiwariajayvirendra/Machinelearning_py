```python
# It support formate AVI insted of mp4
# Aperture is called per second pic 
# Video Capturing 
import cv2

#Webcam open
cam=cv2.VideoCapture(0)

# Video Writer Setup
fourcc=cv2.VideoWriter_fourcc(*'XVID')
out=cv2.VideoWriter(
    r'C:\Users\ajayt\OneDrive\Desktop\DataSets\output.mp4',
    fourcc,
    20.0,
    (640,480)
)

while True:
    ret,frame = cam.read()
    
    if not ret:
        break
    #Write Frame to file
    out.write(frame)

    #Show recording Window
    cv2.imshow('Recording', frame)

    # Press 'Q' to stop
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break
# Release Resources
cam.release()
out.release()   # Important step
cv2.destroyAllWindows()
        
```

```python
import cv2

cam = cv2.VideoCapture(r'C:\Users\ajayt\OneDrive\Desktop\DataSets\traffic.mp4')

# Haarcascade XML file ka exact path dena zaruri hai
car_cascade = cv2.CascadeClassifier(
    r'C:\Users\ajayt\OneDrive\Desktop\ML_Tanay_Sir\cars.xml'
)

count = 0

while True:
    ret, frame = cam.read()
    if not ret:
        break

    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

    # Correct function name
    cars = car_cascade.detectMultiScale(gray, 1.1, 2)

    for (x, y, w, h) in cars:
        cv2.rectangle(frame, (x, y), (x + w, y + h), (0, 255, 0), 2)

    # Count cars per frame
    count += len(cars)

    cv2.putText(frame, 'Vehicle Count: ' + str(count),
                (20, 70), cv2.FONT_HERSHEY_SIMPLEX, 1, (255, 0, 0), 2)

    cv2.imshow('Cars', frame)

    if cv2.waitKey(1) == 27:  # ESC key
        break

cam.release()
cv2.destroyAllWindows()
```

```python
import cv2   # OpenCV library import karna

# Video file open karna (traffic.mp4)
cam = cv2.VideoCapture(r'C:\Users\ajayt\OneDrive\Desktop\DataSets\traffic.mp4')

# Haarcascade XML file load karna (cars.xml)
car_cascade = cv2.CascadeClassifier(
    r'C:\Users\ajayt\OneDrive\Desktop\ML_Tanay_Sir\cars.xml'
)

count = 0   # Vehicle count initialize karna

while True:
    ret, frame = cam.read()   # Video se ek frame read karna
    if not ret:               # Agar frame read nahi hua (video khatam ho gaya)
        break

    # Frame ko grayscale me convert karna (cascade detection ke liye)
    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

    # Cars detect karna Haarcascade se
    cars = car_cascade.detectMultiScale(gray, 1.1, 2)

    # Har detected car ke liye rectangle draw karna
    for (x, y, w, h) in cars:
        cv2.rectangle(frame, (x, y), (x + w, y + h), (0, 255, 0), 2)

    # Current frame me jitni cars detect hui unko count me add karna
    count += len(cars)

    # Frame par vehicle count text likhna
    cv2.putText(frame, 'Vehicle Count: ' + str(count),
                (20, 70), cv2.FONT_HERSHEY_SIMPLEX, 1, (255, 0, 0), 2)

    # Frame ko display karna ek window me
    cv2.imshow('Cars', frame)

    # Agar ESC key (27) press hui to loop break karna
    if cv2.waitKey(1) == 27:
        break

# Video aur windows release/close karna
cam.release()
cv2.destroyAllWindows()
```
