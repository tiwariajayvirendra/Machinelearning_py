```python
import nltk

corpus=""" 
Hello Welcome to k ' Ajay ML Hindi Learning Channel Please do subscribe the channel.
"""
```

```python
### Paragraph ===> Sentaence
from nltk.tokenize import sent_tokenize
```

```python
sent_tokenize(corpus)
```

```python
### Paragraph ===> words
from nltk.tokenize import word_tokenize
word_tokenize(corpus)
```

```python
word_tokenize(corpus)
```

```python
from nltk.tokenize import wordpunct_tokenize
wordpunct_tokenize(corpus)
```

```python
from nltk.tokenize import TreebankWordTokenizer
```

```python
tokenize=TreebankWordTokenizer()
```

```python
tokenize.tokenize(corpus)
```

```python
# Stemming 
#Stemmin is process of reducing a word to its word stem thata 
#affixes to suffixes and prefixes or to the roots of words
#know as a lemma. stemming is important in natural 
#language unsderstandingg (NLU) and natureal language Processing (NLP)
```

```python
from nltk.stem import PorterStemmer
```

```python
stemming = PorterStemmer()
```

```python
words=['eating','eats','eaten','writing','writes','programming','programs','history','finally','finalize']
```

```python
for word in words:
    print(word+"===>"+stemming.stem(word))
```

```python
stemming.stem('congratualtions')
```

```python
### Lancaster Stemming algorithm
from nltk.stem import LancasterStemmer
```

```python
lancaster=LancasterStemmer()
```

```python
for word in words:
    print(word+"====>"+lancaster.stem(word))
    
```

```python
### RegexpStemmer Class
#NLTK Has Regexp Stemmer
#class with the help of which we can easily implement Regular Expression  
# Stemmer algorithms. It bassically takes  asingle regular expression and remover any 
#prefix or suffix that matches the expression let us see an example
```

```python
from nltk.stem import RegexpStemmer
```

```python
reg_stemmer=RegexpStemmer('ing|s$|able$',min=4)
```

```python
reg_stemmer.stem("ingplaying")
```

```python
### Snowball Stemmer
from nltk.stem import SnowballStemmer

```

```python
snowballstemmer=SnowballStemmer('english',ignore_stopwords=False)
```

```python
for word in words:
    print(word+"===>"+snowballstemmer.stem(word))
```

### Wordnet lemmatizer
#  Lemmatizationn technique is like stemming .The output we will get after lematization is called 'lemma',which is a root  word rather thatn root stem. the output of stemming. After lemmatization,we will getting a valid word that means the same thing.

#NLTK Prvides WordNetLemmatizer class which is a thin wrapper around the wordnt corpus .this class uses morphy() function to the WordNet CorpusReader class to find a lemma. let us understant it with example.


```python
import nltk
from nltk.stem import WordNetLemmatizer
lemmatizer= WordNetLemmatizer()
```

```python
for word in words:
    print(word+"===>"+lemmatizer.lemmatize(word,pos='v'))
```

```python
lemmatizer.lemmatize('better',pos='v')
```
