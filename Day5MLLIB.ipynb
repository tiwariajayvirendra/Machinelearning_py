```python
import matplotlib.pyplot as plt
```

```python
#PLT DONUT GRAPH PLOTING

labels=["A","B","C","D","E","F"]
sizes=[30,40,20,50,80,10]
fig,ax=plt.subplots()
ax.pie(sizes,labels=labels,autopct="%1.1f%%")
circle= plt.Circle((0,0),.7,fc='White')   # FC stands For FOr Ground CoLOR
fig.gca().add_artist(circle)

plt.show()
```

```python
# BAR GRAPH PLOTTING 
year = [2022,2023,2024,2025]
production=[1000,1400,1900,2300]
plt.xticks(year)
plt.yticks(production)
plt.bar(year,production)

plt.show()
```

```python
languages= ["Python","Java","C","C++","Go"]
popularity=[90,75,65,85,50]
plt.figure(figsize=(8,5))
plt.bar(languages,popularity)
plt.xlabel("Programming Language")
plt.ylabel("Popularity(%)")

plt.title("Programming Languages Populartiy")
for i , value in enumerate(popularity):
    plt.text(i,value+1,str(value),ha="center")
plt.grid(axis='y')
plt.show
```

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
import seaborn as sns

# Data
data = {"Math":[85,99,78,92],
        "Science":[88,76,85,89],
        "English":[90,85,88,82]}
student = ["Aman","Ajay","Shubham","Ritesh"]

df = pd.DataFrame(data, index=student)

# ✅ Use .values to pass numeric array
plt.imshow(df.values, cmap="coolwarm", aspect='auto')
plt.colorbar()

# Axis labels
plt.xticks(range(len(df.columns)), df.columns)
plt.yticks(range(len(df.index)), df.index)
plt.title("Student Marks HeatMap")

# Annotate values
for i in range(len(df.index)):
    for j in range(len(df.columns)):
        plt.text(j, i, df.iloc[i, j], ha="center", va="center", color="black")

plt.show()
```
