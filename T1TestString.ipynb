```python
# 1. Curly braces (most common)
d1 = {"name": "Ajay", "age": 22}

# 2. dict() constructor
d2 = dict(name="Ajay", age=22)

# 3. Using list of tuples
d3 = dict([("name", "Ajay"), ("age", 22)])

# 4. Using zip()
keys = ["name", "age"]
values = ["Ajay", 22]
d4 = dict(zip(keys, values))

# 5. Using dictionary comprehension
d5 = {x: x**2 for x in range(5)}

# 6. From two lists with loop
keys = ["name", "age"]
values = ["Ajay", 22]
d6 = {}
for k,v in zip(keys, values):
    d6[k] = v

# 7. Using fromkeys()
d7 = dict.fromkeys(["a","b","c"], 0)

# 8. Nested dictionary
d8 = {"student": {"name": "Ajay", "age": 22}}

# 9. Using default values with dict comprehension
d9 = {k: None for k in ["x","y","z"]}

# 10. Using generator expression with dict()
d10 = dict((i, i*2) for i in range(5))

# 11. Using update()
d11 = {}
d11.update({"name":"Ajay", "age":22})

# 12. Using setdefault()
d12 = {}
d12.setdefault("name", "Ajay")
d12.setdefault("age", 22)

# 13. Using JSON string
import json
d13 = json.loads('{"name":"Ajay","age":22}')

# 14. Using dict comprehension with condition
d14 = {i: i*i for i in range(10) if i%2==0}

# 15. Using enumerate()
d15 = {i: v for i,v in enumerate(["Ajay","Kan"])}

# 16. Using dictionary unpacking
d16 = {**{"a":1}, **{"b":2}}

# 17. Using defaultdict (collections)
from collections import defaultdict
d17 = defaultdict(int)
d17["x"] += 1

# 18. Using OrderedDict
from collections import OrderedDict
d18 = OrderedDict([("name","Ajay"),("age",22)])

# 19. Using dict comprehension from string
d19 = {ch: ord(ch) for ch in "abc"}

# 20. Using constructor with keyword + unpack
base = {"x":1}
d20 = dict(base, y=2, z=3)

```
